```python
data=pd.read_csv('/kaggle/input/sms-spam-collection-dataset/spam.csv', encoding='latin-1')
data
```

```python
data = data.rename(columns={'v1': 'Category', 'v2': 'Message'})

# Удаляем лишние колонки (Unnamed: 2, 3, 4)
data = data[['Category', 'Message']]
data.columns
```

```python
data.info()
```

**Dropped The Column Unnamed: 0**

```python
data.isna().sum()
```

```python
data['Spam']=data['Category'].apply(lambda x:1 if x=='spam' else 0)
data.head(5)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(data.Message,data.Spam,test_size=0.25)
```

```python
#CounterVectorizer Convert the text into matrics
from sklearn.feature_extraction.text import CountVectorizer
```

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
from sklearn.pipeline import Pipeline
clf=Pipeline([
    ('vectorizer',CountVectorizer()),
    ('nb',MultinomialNB())
])
```

# Tarining The Model

```python
clf.fit(X_train,y_train)
```

**Here I given Two email Two detect 1st One is looking good and the other one looking spam**

```python
emails=[
    'Sounds great! Are you home now?',
    'Will u meet ur dream partner soon? Is ur career off 2 a flyng start? 2 find out free, txt HORO followed by ur star sign, e. g. HORO ARIES'
]
```

**Predict Email**

```python
clf.predict(emails)
```

# Prediction Of Model

```python
clf.score(X_test,y_test)
```

```python
test_messages = [
    "Hey, are we meeting tomorrow?",  # ham
    "WIN FREE IPHONE NOW!!!",  # spam
    "Can you call me?",  # ham
    "URGENT: Your account compromised",  # spam
    "Meeting moved to 3pm",  # ham
    "Congratulations you won $1000",  # spam
    "What's for dinner?",  # ham
    "TXT STOP to unsubscribe",  # spam
]

predictions = clf.predict(test_messages)
for msg, pred in zip(test_messages, predictions):
    label = "СПАМ" if pred == 1 else "НЕ СПАМ"
    print(f"'{msg}' → {label}")
```
